```python
import requests
import pandas as pd
import os
```

```python
BASE_URL = "http://api.jolpi.ca/ergast/f1"
```

```python
# Hàm gọi API và trả về dữ liệu JSON
def fetch_data(endpoint, year):
    url = f"{BASE_URL}/{year}/{endpoint}" if "{year}" in endpoint else f"{BASE_URL}/{endpoint}"
    response = requests.get(url)
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Failed to fetch data for {endpoint} in year {year}. Status: {response.status_code}")
        return None
```

```python
# Danh sách endpoint và cấu trúc key_path
# endpoints = {
#     "drivers": (["MRData", "DriverTable", "Drivers"], "drivers.csv"),
#     "constructors": (["MRData", "ConstructorTable", "Constructors"], "constructors.csv"),
#     "circuits": (["MRData", "CircuitTable", "Circuits"], "circuits.csv"),
#     "races": (["MRData", "RaceTable", "Races"], "races.csv"),
#     "results": (["MRData", "RaceTable", "Races"], "results.csv"),
#     "driverstandings": (["MRData", "StandingsTable", "StandingsLists"], "driverstandings.csv"),
#     "constructorstandings": (["MRData", "StandingsTable", "StandingsLists"], "constructorstandings.csv"),
#     "qualifying": (["MRData", "RaceTable", "Races"], "qualifying.csv"),
#     "sprint": (["MRData", "RaceTable", "Races"], "sprint.csv"),
#     "status": (["MRData", "StatusTable", "Status"], "status.csv")
# }

endpoints = {
    "seasons": ["MRData", "SeasonTable", "Seasons"],
    "circuits": ["MRData", "CircuitTable", "Circuits"],
    "races": ["MRData", "RaceTable", "Races"],
    "constructors": ["MRData", "ConstructorTable", "Constructors"],
    "drivers": ["MRData", "DriverTable", "Drivers"],
    "results": ["MRData", "RaceTable", "Races"],
    "sprint": ["MRData", "RaceTable", "Races"],
    "qualifying": ["MRData", "RaceTable", "Races"],
    "pitstop": ["MRData", "RaceTable", "Races"],
    "lap": ["MRData", "RaceTable", "Races"],
    "driverstanding": ["MRData", "StandingsTable", "StandingsLists"],
    "constructorstanding": ["MRData", "StandingsTable", "StandingsLists"],
    "status": ["MRData", "StatusTable", "Status"]
}

# Hàm xử lý dữ liệu từng endpoint
def process_data(endpoint, key_path, year_range=range(2000, 2025)):
    for year in year_range:
        print(f"Fetching data for endpoint '{endpoint}' in year {year}...")
        data = fetch_data(endpoint, year)
        if data:
            # Duyệt qua đường dẫn key_path để lấy đúng phần dữ liệu
            current_data = data
            for key in key_path:
                current_data = current_data.get(key, {})
            # Nếu dữ liệu là danh sách, xử lý và lưu file
            if isinstance(current_data, list):
                df = pd.DataFrame(current_data)
                # Tạo thư mục và lưu file theo năm
                os.makedirs(f"data/{endpoint}", exist_ok=True)
                file_path = os.path.join(f"data/{endpoint}", f"{endpoint}_{year}.csv")
                df.to_csv(file_path, index=False)
                print(f"Saved data for endpoint '{endpoint}' in year {year} to {file_path}")
            else:
                print(f"Data for {endpoint} in year {year} is not a list!")


# def process_data(endpoint, key_path, filename):
#     print(f"Fetching data for endpoint '{endpoint}'...")
#     data = fetch_data(endpoint)
#     year = 2025
#     if data:
#         # Duyệt qua đường dẫn key_path để lấy đúng phần dữ liệu
#         current_data = data
#         for key in key_path:
#             current_data = current_data.get(key, {})
#         # Nếu dữ liệu là danh sách, gộp vào danh sách tổng
#         if isinstance(current_data, list):
#             df = pd.DataFrame(current_data)
            
#             # Tạo thư mục 'data' nếu chưa tồn tại
#             os.makedirs("data", exist_ok=True)
#             file_path = os.path.join("data", filename)
#             df.to_csv(file_path, index=False)
#             print(f"Saved data for endpoint '{endpoint}' to {file_path}")
#         else:
#             print(f"Data for endpoint '{endpoint}' is not a list!")
#     else:
#         print(f"No data returned for endpoint '{endpoint}'.")


# Lấy dữ liệu và lưu file CSV cho từng endpoint
for endpoint, key_path in endpoints.items():
    process_data(endpoint, key_path)
```
