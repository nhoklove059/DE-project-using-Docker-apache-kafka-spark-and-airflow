```python
BASE_URL = "http://api.jolpi.ca/ergast/f1"
```

```python
def get_season_data(api=BASE_URL + "/seasons", limit=30):
    all_data = []
    offset = 0
    total = 0
    
    while True:
        url = f"{api}?limit={limit}&offset={offset}"
        response = requests.get(url)

        if response.status_code == 200:
            data = (
                response.json()
                .get("MRData", {})
                .get("SeasonTable", {})
                .get("Seasons", [])
            )
            
            if not data:  
                break
            
            all_data.extend(data)
            offset += limit
            total = int(response.json()["MRData"]["total"])
            print(f"Fetching data {limit}/{offset}...")
            if offset >= total:
                break
            
            df = pd.DataFrame(data)
            os.makedirs("data2/season", exist_ok=True)
            df.to_csv("data2/season/seasons.csv", index=False)
        else:
            print(f"Failed to fetch data for {api}. Status: {response.status_code}")


get_season_data()
```

```python
def get_race_data(year_range = range(2000, 2024)):
    for year in year_range:
        api = f"http://api.jolpi.ca/ergast/f1/{year}/races/"
        response = requests.get(api)
        if response.status_code == 200:
            data = response.json()
            races = data["MRData"]["RaceTable"]["Races"]
            df = pd.DataFrame(races)
            os.makedirs("data2/races", exist_ok = True)
            df.to_csv(f"data2/races/races_{year}.csv", index=False)
        else:
            print(f"Failed to fetch data for {api}. Status: {response.status_code}")
            
get_race_data()
```
