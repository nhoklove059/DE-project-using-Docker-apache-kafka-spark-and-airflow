```python
import requests
from bs4 import BeautifulSoup
```

```python
# url = "https://www.formula1.com/"
# response = requests.get(url)

url = "https://www.thegioididong.com/"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3"
}
response = requests.get(url, headers=headers)
```

```python
if response.status_code == 200:
    soup = BeautifulSoup(response.content, "html.parser")
    products = soup.select(".listproduct > li.item")

    # Extract details for each product
    for product in products:
        try:
            # Extract product name
            name = product.select_one("h3").get_text(strip=True)

            # Extract rating
            rating_element = product.select_one(".vote-txt > b")
            rating = rating_element.get_text(strip=True) if rating_element else "N/A"

            # Extract comment or sold count
            sold_count_element = product.select_one(".rating_Compare span")
            sold_count = (
                sold_count_element.get_text(strip=True) if sold_count_element else "N/A"
            )

            # Print extracted information
            print(f"Product Name: {name}")
            print(f"Rating: {rating}")
            print(f"Comments or Sold Count: {sold_count}")
            print("-" * 40)

        except Exception as e:
            print(f"Error extracting data for a product: {e}")
```
